# Retail Sales Performance Analysis

A Python-based analysis of retail sales data using Pandas and Matplotlib.

The project focuses on understanding sales performance, product performance,
customer activity, discounts, promotions, and sales trends.

**Data source:** `data/store_data.xlsx`

The workbook contains the following tables:

- Sheet3 — Sales Transactions
- Dim Customers — Customer information
- Dim Product — Product information
- Dim Promotion — Promotion information

```python
import sys
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt

sys.path.append("..")

from src.data_loader import load_data
from src.data_cleaning import clean_data
from src.feature_engineering import create_features
from src import analysis, visualization

pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

## Business Insights

The analysis provides several insights into retail sales performance:

1. Monthly sales trends help identify periods of higher and lower sales activity.

2. Product-level analysis highlights the products contributing the most to net sales.

3. Promotion analysis allows different promotional campaigns to be compared based on sales and units sold.

4. Discount analysis helps understand how discount values vary across products.

5. Customer-level analysis identifies customers with higher transaction and sales values.

6. City-level analysis can be used to compare sales performance across different locations.

7. Period comparisons provide a simple way to evaluate changes in sales performance  between two selected time periods.
